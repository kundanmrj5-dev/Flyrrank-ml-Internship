## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/kundanmrj5-dev/Flyrrank-ml-Internship/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

pages = (
    df.loc[(df["impressions_90d"] > 0) & (df["content_age_days"] >= 90)]
      .drop_duplicates("content_id")
      .copy()
)

pages["is_declining_proxy"] = pages["trend_direction"].eq("down")

display(pages[[
    "content_id", "impressions_90d", "sessions_90d",
    "content_age_days", "trend_direction", "is_declining_proxy"
]].head())

print("Rows and columns:", pages.shape)
print("Declining-proxy share:", pages["is_declining_proxy"].mean())
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule uses hand-set cutoffs, while a model can combine several observed signals—such as impressions, average position, page age, and freshness—to rank pages for review. The model is useful only if a fair evaluation shows it improves the review ranking over the fixed-rule baseline. This starter proxy describes the current trend, so it does not prove which pages will decline in the future. A person must review recommendations; the model should not update pages automatically.
